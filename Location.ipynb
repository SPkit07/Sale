```python
import pandas as pd
import numpy as np
import os
import openpyxl
import hashlib
```

```python
folder_path = r'G:\My Drive\KMART\K5\ปี 69\เดือน 03-69'
data = [] 
datafilename = []
bad_files = []
```

```python
def clean_line(value):
    if pd.isna(value): return None
    
    val_str = "".join(str(value).split())
    parts = val_str.split(",")
    if len(parts) >= 5:
        return ",".join(parts[2:-1])
    return None
```

```python
for filename in os.listdir(folder_path):
    if  filename.lower().endswith(".xlsx") and not filename.startswith("~$"):
        filepath = os.path.join(folder_path, filename)
        
        try:
            df = pd.read_excel(filepath ,usecols=[0] ,engine='openpyxl')
            if df.empty :
                bad_files.append({'filename' : filename , 'resons' : "ไฟล์ว่าง"})
                continue

            # ลบค่า NaN
            col = pd.DataFrame(df.iloc[:, 0].dropna())
            # ส่งชื่อไฟล์
            if col.empty :
                bad_files.append({'filename': filename, 'reason': "ไม่มีข้อมูลหลังลบ NaN"})
               
                continue

            col['Fliename'] = filename
            data.append(col)

        except Exception as e:
            bad_files.append({"filename": filename, "reason": f"Error: {str(e)}"})    
        
```

```python
if data:
    final_df = pd.concat(data, ignore_index=True)
    print("รวมข้อมูลสำเร็จ!")
else:
    print("ไม่มีข้อมูลให้รวม ลองเช็ก bad_files ดูครับ")
```
