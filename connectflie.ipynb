```python
import pandas as pd
import glob
import os

# 1. กำหนด Path (ใส่ r ข้างหน้าเพื่อป้องกันปัญหา backslash)
path = r'C:\Users\KS\Desktop\ไฟล์เก็บข้อมูลK1'

# 2. หานามสกุล .xlsx 
all_files = glob.glob(os.path.join(path, "*.xlsx"))


print(f"พบไฟล์ Excel ทั้งหมด: {len(all_files)} ไฟล์")

if len(all_files) == 0:
    print("❌ ยังหาไฟล์ไม่เจอ! ตรวจสอบว่าชื่อโฟลเดอร์ 'ไฟล์เก็บข้อมูลK1' พิมพ์สะกดถูกต้องหรือไม่")
else:
    Merge = []

    
    for filename in all_files:
        # 3. ใช้ pd.read_excel แทน pd.read_csv
        df = pd.read_excel(filename)
        
        # 4. ดึงชื่อไฟล์มาสร้างคอลัมน์วันที่ (ตัด .xlsx ออก)
        file_name_only = os.path.basename(filename).replace('.xlsx', '')
        df['DataDate'] = file_name_only
        
        Merge.append(df)

    # 5. รวมไฟล์ทั้งหมด
    df_final = pd.concat(Merge, axis=0, ignore_index=True)
    
    print("✅ รวมไฟล์สำเร็จ!")
```

```python
col_date = df_final['DataDate']
df_final.insert(0,'DATE',col_date)
```

```python
df_final.drop(columns=['DataDate'],inplace=True)
```

```python
# แปลงคอลัมน์ DATE ให้เป็นวันที่จริงๆ
# dayfirst=True เพราะรูปแบบในรูปคือ วัน-เดือน-ปี
df_final['DATE'] = pd.to_datetime(df_final['DATE'], dayfirst=True).dt.date

# สั่งเรียงลำดับจากน้อยไปมาก
df_final = df_final.sort_values('DATE')
```
