```python
import os 
import pandas as pd 
import numpy as np
import glob
import openpyxl
```

```python
PATHS = {
    "K1": [r"C:\Users\KS\Desktop\SP-DS\K mart 1\เดือน 05",'11'],
    "K2": [r"C:\Users\KS\Desktop\SP-DS\K mart 2\เดือน 05",'21'],
    "K3": [r"C:\Users\KS\Desktop\SP-DS\K mart 3\เดือน 05",'31'],
    "K4": [r"C:\Users\KS\Desktop\SP-DS\K mart 4\เดือน 05",'41'],
    "K5": [r"C:\Users\KS\Desktop\SP-DS\K mart 5\เดือน 05",'51'],
    "SP": [r"C:\Users\KS\Desktop",'00']
}
```

```python

part_a = r'G:\My Drive\KMART(ใช้งาน)'
part_b = r'G:\My Drive\SP'
```

```python
Kmart = str(input("สาขา")).upper()
bill = str(input("เดือน-ท้ายท้ายบิล"))
```

```python
PATHS[f'{Kmart}'][0]
```

```python
if Kmart in PATHS :
    search_a = os.path.join(part_a ,"**" , f"{Kmart}-{bill}.XLSX")
    search_b = os.path.join(part_b ,"**" , f"{Kmart}-{bill}.XLSX")
    file_a = glob.glob(search_a,recursive=True)
    file_b = glob.glob(search_b,recursive=True)
    file = file_a + file_b
    if file :


        df = pd.read_excel(file[0] ,engine='openpyxl' , usecols='A',names=['forBplus'])
        if PATHS[f'{Kmart}'][1] == df['forBplus'].str.split(',').str[1].iloc[0]:

            save_path = os.path.join(PATHS[f'{Kmart}'][0], f"{Kmart}-{bill}.txt")
            
            if os.path.exists(save_path):
                # เปลี่ยนชื่อจาก "K1-05.txt" ให้กลายเป็น "K1-05-1.txt"
                txt_filename = f"{Kmart}-{bill}-1.txt"
                save_path = os.path.join(part_a, txt_filename)
                
            with open(save_path,'w',encoding='utf-8') as f:
                raw_text = df.to_string(index=False, header=False)
            
                # 2. 💡 ตัดช่องว่างส่วนเกินหน้า-หลังของแต่ละบรรทัด เพื่อบังคับให้ชิดซ้ายสุด
                cleaned_lines = [line.strip() for line in raw_text.split('\n')]

                cleaned_text = '\n'.join(cleaned_lines)
                f.write(cleaned_text)
                print("จำนวนแถวทั้งหมด:", len(df))
                print(f'ส่งไฟล์:{Kmart} สำเร็จ : ตำแหน่งไฟล์ {PATHS[Kmart][0]}')

        else :
            print('รหัสสาขาไม่ตรงกัน')
    else:
        print(f'ไม่พบข้อมูล {Kmart}')
else:
    print(f"ไม่พบไฟล์สาขา {Kmart}")
```
